```python
import joblib
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X_train_processed = joblib.load("../models/regression/X_train_processed.pkl")
X_test_processed = joblib.load("../models/regression/X_test_processed.pkl")
y_train = joblib.load("../models/regression/y_train.pkl")
y_test = joblib.load("../models/regression/y_test.pkl")


# Linear Regression
linear_model = LinearRegression()
linear_model.fit(X_train_processed, y_train)
linear_pred = linear_model.predict(X_test_processed)

linear_mae = mean_absolute_error(y_test, linear_pred)
linear_mse = mean_squared_error(y_test, linear_pred)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_pred))
linear_r2 = r2_score(y_test, linear_pred)

print("Linear Regression")
print("MAE:", linear_mae)
print("MSE:", linear_mse)
print("RMSE:", linear_rmse)
print("R²:", linear_r2)

print()

# Decision Tree
tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X_train_processed, y_train)
tree_pred = tree_model.predict(X_test_processed)

tree_mae = mean_absolute_error(y_test, tree_pred)
tree_mse = mean_squared_error(y_test, tree_pred)
tree_rmse = np.sqrt(mean_squared_error(y_test, tree_pred))
tree_r2 = r2_score(y_test, tree_pred)

print("Decision Tree")
print("MAE:", tree_mae)
print("MSE:", tree_mse)
print("RMSE:", tree_rmse)
print("R²:", tree_r2)

print()

# Random Forest
forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
forest_model.fit(X_train_processed, y_train)
forest_pred = forest_model.predict(X_test_processed)

forest_mae = mean_absolute_error(y_test, forest_pred)
forest_mse = mean_squared_error(y_test, forest_pred)
forest_rmse = np.sqrt(mean_squared_error(y_test, forest_pred))
forest_r2 = r2_score(y_test, forest_pred)

print("Random Forest")
print("MAE:", forest_mae)
print("MSE:", forest_mse)
print("RMSE:", forest_rmse)
print("R²:", forest_r2)


results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        tree_mae,
        forest_mae
    ],
    "MSE": [
        linear_mse,
        tree_mse,
        forest_mse
    ],
    "RMSE": [
        linear_rmse,
        tree_rmse,
        forest_rmse
    ],
    "R2": [
        linear_r2,
        tree_r2,
        forest_r2
    ]
})

results

results.sort_values(by="R2", ascending=False)

import joblib
joblib.dump(forest_model,"../models/regression/best_regression_model.pkl")

results.to_csv("../models/regression/regression_results.csv",index=False)
```
