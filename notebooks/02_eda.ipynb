```python
import matplotlib.pyplot as plt 
import seaborn as sns 
import pandas as pd

df = pd.read_csv("../data/processed/cleaned_data.csv")


plt.hist(df['selling_price'], bins=50)
plt.title("Selling Price Distribution")
plt.xlabel("Selling price")
plt.ylabel("Number of cars")
plt.show()


fig, ax = plt.subplots(1,5, figsize=(20,5))

ax[0].scatter(df['vehicle_age'], df['selling_price'])
ax[0].set_title("Vehicle Age vs Selling Price")
ax[0].set_xlabel("Vehicle Age")
ax[0].set_ylabel("Selling Price")

ax[1].scatter(df['km_driven'], df['selling_price'])
ax[1].set_title("km_driven vs Selling Price")
ax[1].set_xlabel("km driven")
ax[1].set_ylabel("Selling Price")

ax[2].scatter(df['max_power'], df['selling_price'])
ax[2].set_title("'max_power' vs Selling Price")
ax[2].set_xlabel("'max_power'")
ax[2].set_ylabel("Selling Price")

ax[3].scatter(df['engine'], df['selling_price'])
ax[3].set_title("enginevs Selling Price")
ax[3].set_xlabel("engine")
ax[3].set_ylabel("Selling Price")

ax[4].scatter(df['mileage'], df['selling_price'])
ax[4].set_title("Mileage vs Selling Price")
ax[4].set_xlabel("Mileage")
ax[4].set_ylabel("Selling Price")

plt.tight_layout()
plt.show()

corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap ='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()


```
