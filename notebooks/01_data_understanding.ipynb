```python
import numpy as np 
import pandas as pd 

df = pd.read_csv("../data/raw/cardekho_dataset.csv")
print(df.head(5))
print(f"Shape:\n{df.shape}\n\n")
print(f"Columns:\n{df.columns}\n\n")
print(df.describe())
print()
df.info()

missing_values = df.isna().sum()
duplicate_rows = df.duplicated().sum()

print("Null values:", missing_values)
print("Duplicate rows:",duplicate_rows)

df.drop(columns=['Unnamed: 0'],inplace=True)
print(f"Columns:\n{df.columns}\n\n")
df.info()

numeric_cols = [
    'vehicle_age',
    'km_driven',
    'mileage',
    'engine',
    'max_power',
    'seats',
    'selling_price'
]

for col in numeric_cols:
    print(f"\n{col}")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Zero:", (df[col] == 0).sum())
    print("Negative:", (df[col] < 0).sum())


str_cols = [
    'car_name',
    'brand',
    'model',
    'seller_type',
    'fuel_type',
    'transmission_type'
    ]

for stg in str_cols:
    df[stg] = df[stg].str.strip().str.lower()

print(df.head(5))

print(df['car_name'].value_counts())

df.to_csv("../data/processed/cleaned_data.csv",index=False)
```
