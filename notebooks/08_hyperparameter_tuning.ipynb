```python
import joblib
import pandas as pd

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# 1. LOAD REGRESSION DATA
# ============================================================

X_train_reg = joblib.load(
    "../models/preprocessing/X_train_processed.pkl"
)

X_test_reg = joblib.load(
    "../models/preprocessing/X_test_processed.pkl"
)

y_train_reg = joblib.load(
    "../models/preprocessing/y_train.pkl"
)

y_test_reg = joblib.load(
    "../models/preprocessing/y_test.pkl"
)


# ============================================================
# 2. RANDOM FOREST REGRESSION - HYPERPARAMETER TUNING
# ============================================================

regression_model = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

regression_params = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

regression_grid = GridSearchCV(
    estimator=regression_model,
    param_grid=regression_params,
    cv=3,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    verbose=1
)

regression_grid.fit(
    X_train_reg,
    y_train_reg
)


# ============================================================
# 3. BEST REGRESSION MODEL
# ============================================================

best_regression_model = regression_grid.best_estimator_

print("Best Regression Parameters:")
print(regression_grid.best_params_)

regression_pred = best_regression_model.predict(
    X_test_reg
)

regression_mae = mean_absolute_error(
    y_test_reg,
    regression_pred
)

regression_mse = mean_squared_error(
    y_test_reg,
    regression_pred
)

regression_rmse = mean_squared_error(
    y_test_reg,
    regression_pred
) ** 0.5

regression_r2 = r2_score(
    y_test_reg,
    regression_pred
)

print("\nTuned Random Forest Regression")
print("MAE:", regression_mae)
print("MSE:", regression_mse)
print("RMSE:", regression_rmse)
print("R²:", regression_r2)


# ============================================================
# 4. LOAD CLASSIFICATION DATA
# ============================================================

X_train_clf = joblib.load(
    "../models/classification/X_train_processed.pkl"
)

X_test_clf = joblib.load(
    "../models/classification/X_test_processed.pkl"
)

y_train_clf = joblib.load(
    "../models/classification/y_train.pkl"
)

y_test_clf = joblib.load(
    "../models/classification/y_test.pkl"
)


# ============================================================
# 5. LOGISTIC REGRESSION - HYPERPARAMETER TUNING
# ============================================================

classification_model = LogisticRegression(
    max_iter=1000
)

classification_params = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["lbfgs"]
}

classification_grid = GridSearchCV(
    estimator=classification_model,
    param_grid=classification_params,
    cv=3,
    scoring="f1_weighted",
    n_jobs=-1,
    verbose=1
)

classification_grid.fit(
    X_train_clf,
    y_train_clf
)


# ============================================================
# 6. BEST CLASSIFICATION MODEL
# ============================================================

best_classification_model = classification_grid.best_estimator_

print("\nBest Classification Parameters:")
print(classification_grid.best_params_)

classification_pred = best_classification_model.predict(
    X_test_clf
)

classification_accuracy = accuracy_score(
    y_test_clf,
    classification_pred
)

classification_precision = precision_score(
    y_test_clf,
    classification_pred,
    average="weighted"
)

classification_recall = recall_score(
    y_test_clf,
    classification_pred,
    average="weighted"
)

classification_f1 = f1_score(
    y_test_clf,
    classification_pred,
    average="weighted"
)

print("\nTuned Logistic Regression")
print("Accuracy:", classification_accuracy)
print("Precision:", classification_precision)
print("Recall:", classification_recall)
print("F1:", classification_f1)


# ============================================================
# 7. SAVE TUNED MODELS
# ============================================================

joblib.dump(
    best_regression_model,
    "../models/regression/tuned_regression_model.pkl"
)

joblib.dump(
    best_classification_model,
    "../models/classification/tuned_classification_model.pkl"
)

print("\nTuned models saved successfully.")
```
