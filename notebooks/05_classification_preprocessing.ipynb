```python
import os
import joblib
import numpy as np
import pandas as pd 

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

df = pd.read_csv("../data/processed/cleaned_data.csv")
df["selling_price"].quantile([0.33, 0.66])

df['price_category'] = pd.cut(
    df['selling_price'], 
    bins=[0,445000,700000,df['selling_price'].max()],
    labels=['Budget', 'Mid-Range', 'Premium']
)

df['price_category'].value_counts()
df.to_csv("../data/processed/classification_data.csv", index=False)


x = df.drop(columns=['selling_price', 'price_category'])
y = df['price_category']

numerical_cols = x.select_dtypes(include='number').columns.tolist()
categorical_cols = x.select_dtypes(include='object').columns.tolist()

# print(numerical_cols)
# print(categorical_cols)

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

processed = ColumnTransformer(
    transformers = [
        (
            "num",
            StandardScaler(),
            numerical_cols
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        )
    ]
)

X_train_processed = processed.fit_transform(x_train)
x_test_processed = processed.transform(x_test)


os.makedirs("../models/classification", exist_ok=True)

joblib.dump(processed,"../models/classification/preprocessor.pkl")
joblib.dump(X_train_processed,"../models/classification/X_train_processed.pkl")
joblib.dump(x_test_processed,"../models/classification/X_test_processed.pkl")
joblib.dump(y_train,"../models/classification/y_train.pkl")
joblib.dump(y_test,"../models/classification/y_test.pkl")

```
